```python
import numpy as np
```

```python
# 3_2_8

sums = {}
diffs = {}
for i in range(1, 7):
    for j in range(1, 7):
        s = i + j
        if s in sums.keys():
            sums[s] += 1 / 36
        else:
            sums[s] =  1 / 36
        diff = i - j
        if diff in diffs.keys():
            diffs[diff] += 1 / 36
        else:
            diffs[diff] = 1 / 36

def expect(xk, pk):
    return np.dot(np.array(list(xk)), np.array(list(pk))).sum()

exp1 = expect(sums.keys(), sums.values())
exp2 = expect(diffs.keys(), diffs.values())

def distribution_sd(sums, diffs):
    im_sums = list(sums)
    im_diffs = list(diffs)

    res = {}
    for el1 in im_sums:
        for el2 in im_diffs:
            for i in range(1, 7):
                for j in range(1, 7):
                    if i + j == el1 and i - j == el2:
                        temp = el1 * el2
                        if temp in res.keys():
                            res[temp] += 1 / 36
                        else:
                            res[temp] = 1 / 36
    return res

sums_diffs = distribution_sd(sums.keys(), diffs.keys())
exp12 = expect(sums_diffs.keys(), sums_diffs.values())
round(exp12 - exp1*exp2)
```
