## 1 Від'ємний біноміальний розподіл

Більш детальна інформація в [документації](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.nbinom.html).

Note: here ξ ~ NBinom(n, p) represents the number of failures to get n successes in Bernoulli trials.

```python
from scipy.stats import nbinom
```

```python
n = 5
p = 0.4
```

### 1.1 Щільність

```python
x = np.arange(-0.5, n*5+1, 0.5)
```

Виведіть значення функції щільності від'ємного біноміального розподілу з параметрами $n, p$ в точках вектора $x$.

```python
pmf = nbinom.pmf(x, n, p)
pmf
```

Зобразіть графічно значення функції щільності від'ємного біноміального розподілу з параметрами $n, p$ в точках вектора $x$.

```python
plt.scatter(x, pmf)
plt.grid()
```

Зобразіть графічно значення функції щільності від'ємного біноміального розподілу з параметрами $n, p$ в точках вектора $x1$.

```python
x1 = np.arange(0, n*5+1)
```

```python
plt.scatter(x1, nbinom.pmf(x1, n, p))
plt.grid()
```

### 1.2 Функція розподілу

Виведіть значення функції розподілу від'ємного біноміального розподілу з параметрами $n, p$ в точках вектора $x$.

```python
cdf = nbinom.cdf(x, n, p)
cdf
```

Зобразіть графічно значення функції розподілу від'ємного біноміального розподілу з параметрами $n, p$ в точках вектора $x$.

```python
plt.plot(x, cdf, drawstyle='steps-post')
plt.show()
```

### 1.3 Випадкові величини

Згенеруйте $100$ випадкових величин з від'ємного біноміального розподілу з параметрами $n, p$.

```python
a = nbinom.rvs(n, p, size = 100)
a
```

## 3 Різні параметри

### 3.1 Біноміальний розподіл

Зобразіть графічно функцію щільності, функцію розподілу та гістограму $10000$ випадкових величин біноміального розподілу з параметрами:
1. $n = 15, p = 0.8$
2. $n = 15, p = 0.2$
3. $n = 20, p = 0.4$
4. $n = 20, p = 0.6$
5. $n = 7, p = 0.5$
6. $n = 25, p = 0.5$
7. $n = 100, p = 0.5$

```python
from scipy.stats import binom
params = [(15, 0.8), (15, 0.2), (20, 0.4), (20, 0.6), (7, 0.5), (25, 0.5), (100, 0.5)]
```

```python
# probability mass functions

for n, p in params:
    x = range(-1, n+4, 1)
    pmf = binom.pmf(x, n, p)
    plt.scatter(x, pmf)
    plt.grid()
    plt.title(f'Binom: (n, p) = ({n}, {p})')
    plt.show()
```

```python
# cumulative distribution function

for n, p in params:
    x = range(-1, n+4, 1)
    cdf = binom.cdf(x, n, p)
    plt.plot(x, cdf, color='r', drawstyle='steps-post')
    plt.title(f'Binom: (n, p) = ({n}, {p})')
    plt.show()
```

```python
for n, p in params:
    rvs = binom.rvs(n, p, size=10_000)
    plt.hist(rvs, bins='auto', color='g')
    plt.title(f'Binom: (n, p) = ({n}, {p})')
    plt.show()
```

### 3.2 Геометричний розподіл

Зобразіть графічно функцію щільності, функцію розподілу та гістограму $10000$ випадкових величин геометричного розподілу з параметрами:
1. $p = 0.1$
2. $p = 0.2$
3. $p = 0.3$

5. $p = 0.5$
6. $p = 0.6$
7. $p = 0.7$
8. $p = 0.8$
9. $p = 0.9$

```python
from scipy.stats import geom
```

```python
# probability mass function

params = [0.1*i for i in range(1, 10)]
for p in params:
    x = range(0, int((1-p)/(p**2))+7)
    pmf = geom.pmf(x, p)
    plt.title(f'Geom: (p) = ({p})')
    plt.scatter(x, pmf)
    plt.grid()
    plt.show()
```

```python
# cumulative distribution function

for p in params:
    x = range(0, int((1-p)/(p**2))+10)
    cdf = geom.cdf(x, p)
    plt.title(f'Geom: (p) = ({p})')
    plt.plot(x, cdf, drawstyle='steps-post', color='r')
    plt.show()
```

```python
# random variables

for p in params:
    rvs = geom.rvs(p, size=10_000)
    plt.hist(rvs, bins='auto', color='g')
    plt.title(f'Geom: (p) = ({p})')
    plt.show()
```

### 3.3 Розподіл Пуассона

Зобразіть графічно функцію щільності, функцію розподілу та гістограму $10000$ випадкових величин розподілу Пуассона з параметрами:

## 4 Апроксимація біноміального розподілу розподілом Пуассона

Покажіть схожість біноміального розподілу та розподілу Пуассона (їх функцій щільності) з відповідними параметрами.
Наприклад:
- $n = 100, p = 0.05$
- $n = 100, p = 0.1$
- $n = 100, p = 0.15$
- $n = 100, p = 0.2$
- $n = 100, p = 0.35$
- $n = 100, p = 0.5$
- $n = 100, p = 0.75$

We will use the Poisson limit theorem, which states:

Let $\xi_n$ be a sequence of discrete random variables such that $\xi_n \sim \text{Binom}(n, p_n)$ and the sequence $np_n$ converges to finite limit $\lambda$ as $n \rightarrow \infty$. Then $\lim_{n \to \infty} P(\xi_n = k) = e^{-\lambda} \frac{\lambda^k}{k!}$.

For every pair of parameters (n, p) we will generate a finite, but big sequence of $\xi_n$ which satisfy the theorem conditions. Then, we construct the plots of probability mass functions of $\xi_n$ and $Poisson(\lambda)$. What we expect to see, is the pointwise convergence to the probability mass function of $Poisson(\lambda)$.

```python
params = [(100, i) for i in [0.05, 0.1, 0.15, 0.2, 0.35, 0.5, 0.75]]
```

```python
def rv_sequence(n, p_n):
    for i in range(10, 2*n):
        if i == n:
            continue
        yield (i, n*p_n/i)

for n, p in params:
    x = range(-1, n+3)
    pmf_binom = binom.pmf(x, n, p)
    pmf_poisson = poisson.pmf(x, n*p)
    plt.grid()
    plt.scatter(x, pmf_binom, c='r', label=f"Binom({n}, {p})")
    plt.scatter(x, pmf_poisson, s=15, c='g', label=f"Poisson({n*p})")
    for _n, _p in rv_sequence(n, p):
        pmf_binom = binom.pmf(x, _n, _p)
        plt.scatter(x, pmf_binom, c='black', s = 0.2)
    plt.xlabel("Im(ξ_n) ⊂ R")
    plt.ylabel("P(ξ=k)")
    plt.legend()
    plt.show()
```

```python
# more close view
for n, p in params:
    x = range(max(0, int(n*p)-15), int(n*p)+15)
    pmf_binom = binom.pmf(x, n, p)
    pmf_poisson = poisson.pmf(x, n*p)
    plt.grid()
    plt.scatter(x, pmf_binom, c='r', label=f"Binom({n}, {p})")
    plt.scatter(x, pmf_poisson, s=15, c='g', label=f"Poisson({n*p})")
    for _n, _p in rv_sequence(n, p):
        pmf_binom = binom.pmf(x, _n, _p)
        plt.scatter(x, pmf_binom, c='black', s = 0.2)
    plt.xlabel("Im(ξ_n) ⊂ R")
    plt.ylabel("P(ξ=k)")
    plt.legend()
    plt.show()
```
