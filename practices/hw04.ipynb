```python
def probs(i, j):
    if i == 0:
        if 5 - j < 0:
            return 0
        return (comb(36, 5-j) * comb(12, j)/comb(52, 5))
        
    elif i == 1:
        if j - 1 >= 0 and 5 - j >= 0 and j >= 0 and 4 - j >= 0:
            return (comb(12, j-1) * comb(36, 5-j) + 3 * comb(12, j)*comb(36, 4-j))/comb(52, 5)
        elif j - 1 >= 0 and 5 - j >= 0:
            return (comb(12, j-1) * comb(36, 5-j))/comb(52, 5)
        elif j >= 0 and 4 - j >= 0:
            return (3*comb(12, j)*comb(36, 4-j))/comb(52, 5)
        return 0
        
    elif i == 2:
        if j >= 0 and 3 - j >= 0 and j - 1 >= 0 and 4 - j >= 0:
            return (3*comb(12, j-1) * comb(36, 4-j) + comb(3, 2) * comb(12, j)*comb(36, 3-j))/comb(52, 5)
        elif j - 1 >= 0 and 4 - j >= 0:
            return (comb(3, 1) * comb(12, j-1)*comb(36, 4-j))/comb(52, 5)
        elif j >= 0 and 3 - j >= 0:
            return (comb(3,2)*comb(12, j)*comb(36, 3-j))/comb(52, 5)
        return 0
        
    elif i == 3:
        if j >= 0 and 2 - j >= 0 and j - 1 >= 0 and 3 - j >= 0:
            return (comb(12, j) * comb(36, 2-j) + comb(3, 2) * comb(12, j-1)*comb(36, 3-j))/comb(52, 5)
        elif j - 1 >= 0 and 3 - j >= 0:
            return (comb(3, 2) * comb(12, j-1)*comb(36, 3-j))/comb(52, 5)
        elif j >= 0 and 2 - j >= 0:
            return (comb(12, j)*comb(36, 2-j))/comb(52, 5)
        return 0
        
    elif i == 4:
        if j - 1 >= 0 and 2 - j >= 0:
            return comb(12, j-1) * comb(36, 2-j) / comb(52, 5)
        return 0
    else:
        return 0
```

```python
general = 0
for i in [0, 1, 2, 3, 4]:
    res = []
    for j in range(6):
        res.append(probs(i, j))
    print(res, sum(res))
    general += sum(res)
print(general)
```
