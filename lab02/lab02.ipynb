```python
import pandas as pd
from matplotlib import pyplot as plt

res = []
p = 0.5
for n in [10, 50, 100, 250, 500, 1000, 2500, 5000, 10000]:
    a = (np.random.rand(n) < p).astype(int)
    res.append((n, a.mean()))

df = pd.DataFrame(res, columns = ['n', 'p_emp'])
df
```

```python
plt.plot(df['n'].values, df['p_emp'].values)
plt.plot(df['n'].values, [0.5] * len(df))
```

```python
# 0 for head
# 1 for tail
import pandas as pd
from matplotlib import pyplot as plt

res = []
p = 0.5
for n in [10, 50, 100, 250, 500, 1000, 2500, 5000, 10000, 50000, 100000]:
    tosses = np.random.randint(0, 2, size=n)
    heads = list(tosses).count(0)
    res.append((n, heads / n))

df = pd.DataFrame(res, columns = ['n', 'p_emp'])
df
```

```python
plt.plot(df['n'].values, df['p_emp'].values)
plt.plot(df['n'].values, [0.5] * len(df))
```
