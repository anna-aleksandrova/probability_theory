```python
# 3_1_17
from scipy.stats import binom
from math import comb
def own(n, m, p, i):
    res = 0
    for j in range(1, i):
        res += comb(n, j)*comb(m, i - j) * (p ** i) * (1-p) ** (n+i-2*j)
    return res

m = 2
n = 2
p = 0.5
x1 = []
y1 = []
for i in range(2, m+n+1):
    x1.append(i)
    y1.append(own(m, n, p, i))
pmf = binom.pmf(x1, m+n,p)
print(pmf)
print(y1)
```
