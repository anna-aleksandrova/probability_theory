```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon

l = 1
n = 100000
ksi1 = expon(0, 1)
ksi2 = expon(0, 1)

ksi1_rvs = np.array(ksi1.rvs(size=n))
ksi2_rvs = np.array(ksi2.rvs(size=n))

eta_rvs = ksi1_rvs - ksi2_rvs
fig = plt.figure(figsize=(20, 5))

x = np.arange(-10, 10, 0.01)

def density(x):
    return np.exp(-np.abs(x))/2
    
plt.subplot(1, 2, 1, ylim=(-0.5, 1.5))
plt.hist(eta_rvs, density=True, bins="auto")
pdf_eta = [density(i) for i in x]
plt.plot(x, pdf_eta, c='r')

def cdf(x):
    global l
    if x <= 0:
        return np.exp(l*x)/2
    else:
        return 1-np.exp(-l*x)/2
        
plt.subplot(1, 2, 2, ylim=(-0.5, 1.5))
cdf_eta = [cdf(i) for i in x]
plt.plot(x, cdf_eta, c='r')
plt.show()
```

```python
l = 1
n = 100000
ksi1 = expon(0, 1/l)
ksi2 = expon(0, 1/l)

ksi1_rvs = np.array(ksi1.rvs(size=n))
ksi2_rvs = np.array(ksi2.rvs(size=n))

zeta_rvs = np.abs(ksi1_rvs - ksi2_rvs)
fig = plt.figure(figsize=(20, 5))

x = np.arange(-10, 10, 0.01)

def density(x):
    global l
    if x >= 0:
        return l*np.exp(-l*x)
    else: 
        return 0
    
plt.subplot(1, 2, 1, ylim=(-0.5, 1.5))
plt.hist(zeta_rvs, density=True, bins="auto")
pdf_zeta = [density(i) for i in x]
plt.plot(x, pdf_zeta, c='r')
plt.plot(np.array([1/l for _ in range(1000)]), np.array([x*0.01 for x in range(-500, 500)]))

# def cdf(x):
#     global l
#     if x <= 0:
#         return np.exp(l*x)/2
#     else:
#         return 1-np.exp(-l*x)/2
        
# plt.subplot(1, 2, 2, ylim=(-0.5, 1.5))
# cdf_eta = [cdf(i) for i in x]
# plt.plot(x, cdf_eta, c='r')
plt.show()
```

```python
from scipy.stats import norm

l = 1
n = 100000
ksi1 = norm(0, l)
ksi2 = expon(0, l)

ksi_rvs = np.array(ksi1.rvs(size=n))
eta_rvs = np.array(ksi2.rvs(size=n))

zeta_rvs = eta_rvs + (ksi_rvs**2)/(2*l)
fig = plt.figure(figsize=(20, 5))

x = np.arange(-10, 10, 0.01)

def density(x):
    global l
    if x >= 0:
        return 2*l*np.sqrt(l*x/np.pi)*np.exp(-l*x)
    else: 
        return 0
    
plt.subplot(1, 2, 1, ylim=(-0.5, 1.5))
plt.hist(zeta_rvs, density=True, bins="auto")
pdf_zeta = [density(i) for i in x]
plt.plot(x, pdf_zeta, c='r')
# plt.plot(np.array([1/l for _ in range(1000)]), np.array([x*0.01 for x in range(-500, 500)]))

# def cdf(x):
#     global l
#     if x <= 0:
#         return np.exp(l*x)/2
#     else:
#         return 1-np.exp(-l*x)/2
        
# plt.subplot(1, 2, 2, ylim=(-0.5, 1.5))
# cdf_eta = [cdf(i) for i in x]
# plt.plot(x, cdf_eta, c='r')
plt.show()
```
